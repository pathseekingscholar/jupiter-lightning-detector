**Goal:** demonstrate archive access, calibrated-image ingestion, a persistent candidate catalog, and reproduction of the six detections in Dyudina et al. (2004).

```python
from pathlib import Path
import sqlite3
import jupiter_pipeline as jp

jp.run_all()
```

## Ground-truth detections

The paper supplies exact image IDs and pixel locations. The printed pair is treated as displayed image `(x, y)`, while NumPy reads `[y-1, x-1]`.

```python
with sqlite3.connect(jp.DB_PATH) as con:
    rows = con.execute('''
        SELECT o.image_number, o.start_time, c.label, c.x, c.y,
               ROUND(c.peak_snr, 1) AS peak_snr, c.bright_pixel_count
        FROM candidates c JOIN observations o USING (opus_id)
        ORDER BY o.start_time, c.label
    ''').fetchall()
rows
```

## What this establishes

- OPUS metadata and calibrated products can be acquired programmatically.
- Calibrated 32-bit I/F images can be read directly from the VICAR/PDS files.
- Published lightning locations are reproduced as local multi-pixel enhancements.
- Observations and candidate measurements are stored in SQLite for later searches.
